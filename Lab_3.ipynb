```python
# -------------------------------
# A. Handle Missing Data
# -------------------------------

import numpy as np
import pandas as pd
```

```python
# Create data with missing values
data = {
    'A': [1, 2, np.nan],
    'B': [4, np.nan, np.nan],
    'C': [5, np.nan, np.nan],
    'D': [6, np.nan, np.nan]
}
```

```python
# Convert dictionary into DataFrame
df = pd.DataFrame(data)
```

```python
# Display missing data locations
print("MISSING DATA:")
```

```python
print(df.isna())
```

```python
# Remove rows containing missing values
df_clean = df.dropna()
```

```python
# Replace missing values with 0
df_filled = df.fillna(0)
```

```python
# Display cleaned data
print("\nDATA AFTER REMOVING MISSING VALUES:")
print(df_clean)
```

```python
# Display filled data
print("\nDATA AFTER FILLING MISSING VALUES:")
print(df_filled)
```

```python
# -------------------------------
# B. Data Transformation
# -------------------------------

df = pd.DataFrame({
    'gender': ['male', 'female', 'male']
})
```

```python
# Convert gender into numerical values
df['gender_num'] = df['gender'].map({
    'male': 0,
    'female': 1
})
```

```python
# Rename column
df_renamed = df.rename(columns={
    'gender': 'Gender'
})
```

```python
print("\nDATA TRANSFORMATION:")
print(df_renamed)
```

```python
# -------------------------------
# C. Outlier Detection
# -------------------------------

# Create 100x3 random numerical dataset
data = pd.DataFrame(np.random.randn(100, 6))
```

```python
# Identify values greater than 6 or less than -6
outliers = (data > 6) | (data < -6)
```

```python
# Display number of outliers
print("\nNUMBER OF OUTLIERS IN EACH COLUMN:")
print(outliers.sum())
```
