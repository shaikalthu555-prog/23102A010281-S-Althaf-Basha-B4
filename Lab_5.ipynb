# Name Of The Experiment : Data Visualization with Matplotlib and Seaborn

```python
# Import matplotlib for plotting graphs
import matplotlib.pyplot as plt
```

```python
# A. Line Plot
# ---------------------------

# X-axis values
x = [1, 2, 3, 4]
```

```python
# Y-axis values
y = [10, 20, 25, 30]
```

```python
# Create a line plot using x and y values
plt.plot(x, y)
```

```python
# Add a title to the graph
plt.title('Line Plot')
```

```python
# Set label for X-axis
plt.xlabel('X-axis')
```

```python
# Set label for Y-axis
plt.ylabel('Y-axis')
```

```python
# Display grid lines
plt.grid(True)
```

```python
# Save the plot as an image file
plt.savefig('line_plot.png')
plt.close()
```

```python
# ---------------------------
# B. Bar Plot
# ---------------------------

# Import pandas for DataFrame creation
import pandas as pd
```

```python
# Create sample categorical data
df = pd.DataFrame({
    'Category': ['A', 'B', 'C'],
    'Values': [10, 20, 15]
})
```

```python
# Create a bar chart
df.plot(kind='bar', x='Category', y='Values')
```

```python
# Add chart title
plt.title('Bar Plot')
```

```python
# Save the plot
plt.savefig('bar_plot.png')
plt.close()
```

```python
# ---------------------------
# C. Histogram and Density Plot
# ---------------------------

# Import seaborn for statistical visualization
import seaborn as sns
```

```python
# Import numpy for random data generation
import numpy as np
```

```python
# Generate 1000 random values
data = np.random.randn(1000)
```

```python
# Create histogram with KDE curve
sns.histplot(data, kde=True)
```

```python
# Add title
plt.title('Histogram and Density')
```

```python
# Save figure
plt.savefig('histogram_density.png')
plt.close()
```

```python
# ---------------------------
# D. Scatter Plot
# ---------------------------

# Generate random x values
x = np.random.rand(50)
```

```python
# Generate random y values
y = np.random.rand(50)
```

```python
# Create scatter plot
plt.scatter(x, y)
```

```python
# Add title and labels
plt.title('Scatter Plot')
plt.xlabel('x')
plt.ylabel('y')
```

```python
# Save figure
plt.savefig('scatter_plot.png')
plt.close()
```

```python
# ---------------------------
# E. Box Plot
# ---------------------------

# Create box plot from random data
sns.boxplot(data=np.random.randn(100, 4))
```

```python
# Add chart title
plt.title('Box Plot')
```

```python
# Save figure
plt.savefig('box_plot.png')
plt.close()
```
